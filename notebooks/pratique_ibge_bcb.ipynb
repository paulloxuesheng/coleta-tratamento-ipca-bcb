```python
# Importando bibliotecas
import requests
import pandas as pd
```

```python
# Requisição à API do BCB
URL = "https://api.bcb.gov.br/dados/serie/bcdata.sgs.433/dados/ultimos/20?formato=json"
response = requests.get(URL,timeout=10) #Guardando o resultado na variável response.
print (response.status_code)
```

```python
# Convertendo JSON para DataFrame
dados = response.json()
df = pd.DataFrame(dados)
df.head()
```

```python
# Salvando os dados brutos em CSV
df.to_csv("../data/raw/bcb_tabela.csv", index=False) # Salva sem a coluna de índice
```

```python
# Verificação inicial dos dados
print(df.dtypes) # Para identificar os tipos de colunas (int, str, object,float, ou datetime).
print(df.isnull().sum()) # Para procurar por células vazias e se vazias fazer a soma delas.
print(df.duplicated().sum()) # Para verificar se existem linhas repetidas e se tiver fazer a soma delas.
```

```python
# Tratamento dos Dados.
df["valor"] = pd.to_numeric(df["valor"]) # Converterndo a coluna valor(str) em números reais.
df["data"] = pd.to_datetime(df["data"], format="%d/%m/%Y") # Convertendo a coluna data(str) em data real.
print(df.dtypes)
```

```python
# Salvando os dados tratados em CSV e Parquet
df.to_csv("../data/processed/dados_tratados.csv", index=False)
df.to_parquet("../data/processed/dados_tratados.parquet", index=False)
```

## Respostas Teóricas

### 1. Importância do Web Scraping

O web scraping é útil e importante para podermos automatizar a coleta de dados soltos em sites públicos na internet. Ao invés de fazermos a conferência e coletar manualmente de forma constante ou diária quando os dados são atualizados, com a codificação correta conseguimos automatizar esse processo, transformando informações soltas em tabelas organizadas para análise e tomada de decisão.


### 2. Riscos Legais, Éticos e Medidas Práticas
Para fazer uso de dados públicos é preciso ter bom senso, ética e responsabilidade: ler os termos de uso do site e das informações que serão coletadas, respeitar direitos autorais e seguir à risca a LGPD (aplicando anonimização quando necessária).

Duas medidas práticas importantes são:
- Ler o arquivo `robots.txt` antes de fazer o scraping e não coletar dados pessoais (ou mascará-los/anonimizá-los caso seja estritamente necessário).
- Limitar a velocidade das requisições (inserindo pausas entre os pedidos) para não sobrecarregar nem derrubar o servidor do site.

### 3. Diferenças entre Requisição via API e Sem API (e qual é mais adequada)
- **Via API:** O código acessa diretamente um endereço oficial (URL) disponibilizado pelo servidor e recebe os dados já limpos e estruturados em formatos como JSON ou CSV, sem precisar ler a página visual (HTML), o que torna o trabalho muito mais rápido e fácil.
- **Sem API (Web Scraping):** Não há uma entrega pronta de dados estruturados. É preciso vasculhar o código HTML da página visual para localizar onde a informação está escondida (em tabelas, textos ou botões). O grande problema é que qualquer mudança na estrutura da página — como mover um botão de lugar ou atualizar o layout — quebra o código e faz o robô parar de funcionar.

**Qual é mais adequada?**
Sempre que possível, a mais adequada é a coleta **via API oficial**, pois é mais estável, organizada, rápida e legalmente segura. A coleta **sem API (web scraping)** só deve ser usada como alternativa quando o site não oferece uma API pública.
